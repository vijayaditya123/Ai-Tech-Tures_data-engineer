```python
import pandas as pd 
df= pd.read_csv("ipl.csv" ,low_memory = False)
```

```python
avg_runs = (
    df.groupby(["season","match_id","batting_team","over"], as_index=False)["runs_total"]
      .sum()
      .groupby(["batting_team","over"], as_index=False)["runs_total"]
      .mean()
)


avg_runs
```

```python
import matplotlib.pyplot as plt

pivot = avg_runs.pivot(
    index="batting_team",
    columns="over",
    values="runs_total"
)

plt.figure(figsize=(14, 8))
plt.imshow(pivot, aspect="auto")
plt.colorbar(label="Average Runs")

plt.xlabel("Over")
plt.ylabel("Batting Team")
plt.title("Average Runs per Over for All Teams")

plt.xticks(range(len(pivot.columns)), pivot.columns)
plt.yticks(range(len(pivot.index)), pivot.index)

plt.show()
```
